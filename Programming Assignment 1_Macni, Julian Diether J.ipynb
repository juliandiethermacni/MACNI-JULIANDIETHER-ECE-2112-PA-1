## A. WORD ROTATION PROBLEM

```python
def rotate_word(text): #defines the function and variable
    return text[1:] + text[0] #starts from index 1 to the end, then puts index 0 after 

#prints the expected outputs
print(rotate_word("python"))
print(rotate_word("logic"))
print(rotate_word("Code"))
print(rotate_word("A"))
```

## B. USERNAME BUILDER PROBLEM

```python
def make_username(first_name, last_name): #defines the function and variables
    #converts both first and last name in lowercase, removes all spaces, and joins them with a period
    return (first_name + "." + last_name).lower().replace(" ","")

#prints the expected outputs
print(make_username("Ada", "Lovelace"))
print(make_username("Alan", "Turing"))
print(make_username("Ana Maria", "De Leon"))
```

## C. BOOKEND SWAP PROBLEM

```python
def swap_bookends(items): #defines the function and variable
    first, *middle, last = items #declares the first, middle, and last elements in the list
    return [last] + middle + [first] #swaps the first and last element

#prints the expected outputs
print(swap_bookends([1, 2, 3, 4, 5, 6]))
print(swap_bookends(["red", "green", "blue"]))
print(swap_bookends([8, 3]))
```
